# **SchoolCheck**
## What affects students' academic performance and dropout risk?

### 1. Introduction

#### 1.1 Problem significance
Young people's education is one of the most important factors affecting politics, economy and world's future years. However, dropout remains one of the major problems that the university system has to face every year. According to recent survey, around one in three students across Europe start a degree and dod not finish it[[1]](#ref1). There are many reasons why students may struggle to complete or even start their education. These reasons vary depending on age, nationality and so other factors. <br>
Such a global problem sends a clear message to all educational institutions: reducing university dropout rates must be set as a target aim, as high attrition also reduces teaching institution's sustainability. To address this significant issue, the first step is to identify the major causes of student dropout. Which of them have the greatest impact on student performance?

In this project I am investigating how the combination of different life, psychological and health factors affects univeristy dropout rates. 

#### 1.2 Target group 
SchoolCheck explores datasets containing statistical information related to students' lives in college. Such analytics and graphical representations of the issue may be helpful for future educators who want to understand the core struggles of an average learner and the main factors that may lead to future dropout among incoming students in order to know how to prevent it. Examining this problem may also be valuable for universities which struggle to identify the main drivers of student dropout rates, enabling them to improve the learning environment in order to support better academic performance.
